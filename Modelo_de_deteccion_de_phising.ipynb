```python
# %%
# 1. Instalación de librerías necesarias en Google Colab
# XGBoost ya viene instalado en Colab (si faltara: !pip install xgboost -q)

# 2. Conectar Google Drive a Colab
from google.colab import drive
drive.mount('/content/drive')

import os, time, warnings, numpy as np, pandas as pd, re
from urllib.parse import urlparse
from itertools import product
from scipy.stats import wilcoxon, chi2
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, precision_score, recall_score, roc_auc_score, matthews_corrcoef, confusion_matrix
warnings.filterwarnings('ignore')

try:
    from xgboost import XGBClassifier; HAS_XGB=True
except Exception:
    from sklearn.ensemble import HistGradientBoostingClassifier; HAS_XGB=False
    print('AVISO: sin xgboost; usando HistGradientBoosting solo para prueba')

# Configuración del experimento
QUICK   = False         # False: experimento completo de 10 repeticiones; True: prueba rápida
# Ruta exacta apuntando a tu carpeta en Google Drive
CSV     = '/content/drive/MyDrive/Excel url pishing/malicious_phish.csv'

N_SAMPLE= 6000 if QUICK else 40000
N_REP   = 2    if QUICK else 10
POP     = 4    if QUICK else 8
IT_CASC = 2    if QUICK else 7     # evaluaciones por etapa = POP*(IT+1) -> 64
IT_SOLO = 5    if QUICK else 15    # POP*(IT+1) -> 128
N_RAND  = 8    if QUICK else 64

# Carpeta donde se guardarán los resultados dentro de Colab
os.makedirs('resultados', exist_ok=True)

# %%
df = pd.read_csv(CSV)
print(df.shape, df['type'].value_counts().to_dict())
df = df.drop_duplicates(subset='url')
df = df[df['type'].isin(['benign','phishing'])].copy()
df['y'] = (df['type']=='phishing').astype(int)
print('Tras dedup y binarización:', df.shape, df['y'].mean().round(3))
# muestra estratificada (se conserva la proporción natural de clases)
df, _ = train_test_split(df, train_size=min(N_SAMPLE,len(df)), stratify=df['y'], random_state=0)
df = df.reset_index(drop=True)

# %%
SUSP = ['login','secure','account','update','verify','bank','paypal','signin','confirm','webscr']

def feats(u):
    u = str(u)
    # Manejo seguro de urlparse para evitar "ValueError: Invalid IPv6 URL"
    try:
        p = urlparse(u if '://' in u else 'http://' + u)
        h = p.netloc
        path = p.path
        query = p.query
    except Exception:
        p = None
        h = u.split('/')[0] if '/' in u else u
        path = '/' + '/'.join(u.split('/')[1:]) if '/' in u else ''
        query = ''

    return dict(
      len_url=len(u), len_host=len(h), len_path=len(path), len_query=len(query),
      n_dots=u.count('.'), n_hyphen=u.count('-'), n_under=u.count('_'), n_slash=u.count('/'),
      n_at=u.count('@'), n_qm=u.count('?'), n_eq=u.count('='), n_amp=u.count('&'), n_pct=u.count('%'),
      n_digits=sum(c.isdigit() for c in u), n_letters=sum(c.isalpha() for c in u),
      ratio_digits=sum(c.isdigit() for c in u)/max(len(u),1),
      has_ip=int(bool(re.search(r'(\d{1,3}\.){3}\d{1,3}', h))),
      https=int(u.lower().startswith('https')),
      n_subdomains=max(h.count('.')-1, 0),
      n_susp=sum(w in u.lower() for w in SUSP),
      tld_len=len(h.split('.')[-1]) if '.' in h else 0,
      has_port=int(':' in h),
      path_depth=path.count('/'),
      n_special=sum(not c.isalnum() for c in u)
    )

X = pd.DataFrame([feats(u) for u in df['url']]); y = df['y'].values
print(X.shape)

# %%
NAMES=['learning_rate','max_depth','n_estimators','subsample','colsample_bytree']
def decode(v):
    v=np.clip(v,0,1)
    return dict(learning_rate=float(10**(-2+1.5*v[0])),            # 0.01 – 0.316
                max_depth=int(round(3+7*v[1])),                     # 3 – 10
                n_estimators=int(round(50+250*v[2])),               # 50 – 300
                subsample=float(0.5+0.5*v[3]), colsample_bytree=float(0.5+0.5*v[4]))

def make(p, seed):
    if HAS_XGB:
        return XGBClassifier(**p, eval_metric='logloss', tree_method='hist', n_jobs=-1, random_state=seed)
    return HistGradientBoostingClassifier(learning_rate=p['learning_rate'], max_depth=p['max_depth'],
                                          max_iter=p['n_estimators'], random_state=seed)

def default_model(seed):
    return XGBClassifier(eval_metric='logloss', tree_method='hist', n_jobs=-1, random_state=seed) if HAS_XGB \
        else HistGradientBoostingClassifier(random_state=seed)

class Fit:                       # fitness = 1 - F1 en validación; cuenta evaluaciones (EV)
    def __init__(s,Xtr,ytr,Xva,yva,seed): s.a=(Xtr,ytr,Xva,yva); s.seed=seed; s.ev=0; s.hist=[]; s.best=1
    def __call__(s,v):
        Xtr,ytr,Xva,yva=s.a; m=make(decode(v),s.seed).fit(Xtr,ytr)
        f=1-f1_score(yva,m.predict(Xva)); s.ev+=1; s.best=min(s.best,f); s.hist.append(s.best); return f

def gwo_np(f,lo,hi,pop,iters,rng):
    D=len(lo); X=rng.uniform(lo,hi,(pop,D)); F=np.array([f(x) for x in X])
    for t in range(iters):
        o=np.argsort(F); lead=[X[o[0]].copy(),X[o[1]].copy(),X[o[2]].copy()]; a=2-2*t/iters
        for i in range(pop):
            new=np.zeros(D)
            for L in lead:
                A=2*a*rng.random(D)-a; C=2*rng.random(D); new+=L-A*np.abs(C*L-X[i])
            X[i]=np.clip(new/3,lo,hi); F[i]=f(X[i])
    b=np.argmin(F); return X[b],F[b]

def pso_np(f,lo,hi,pop,iters,rng,w=0.7,c1=1.5,c2=1.5):
    D=len(lo); X=rng.uniform(lo,hi,(pop,D)); V=np.zeros((pop,D)); F=np.array([f(x) for x in X])
    P=X.copy(); PF=F.copy(); g=P[np.argmin(PF)].copy(); gf=PF.min()
    for t in range(iters):
        for i in range(pop):
            V[i]=w*V[i]+c1*rng.random(D)*(P[i]-X[i])+c2*rng.random(D)*(g-X[i])
            X[i]=np.clip(X[i]+V[i],lo,hi); F[i]=f(X[i])
            if F[i]<PF[i]: P[i]=X[i]; PF[i]=F[i]
            if F[i]<gf: g=X[i].copy(); gf=F[i]
    return g,gf

# ---- GWO y PSO implementados en NumPy (sin librerías externas) ----
gwo=gwo_np   # alias
pso=pso_np   # alias

def ic_iter(hist,start,end,pop):
    """Iteración (0 = población inicial) en que se alcanza el mejor valor de la etapa."""
    target=hist[end-1]
    for i in range(start,end):
        if hist[i]==target: return max(int(np.ceil((i-start+1)/pop))-1,0)
    return 0

def cascade(f,pop,iters,rng):
    D=5; lo,hi=np.zeros(D),np.ones(D)
    g,_=gwo(f,lo,hi,pop,iters,rng); ev1=f.ev
    f1_gwo=1-f.best; ic_gwo=ic_iter(f.hist,0,ev1,pop)
    lo2,hi2=np.clip(g-0.2,0,1),np.clip(g+0.2,0,1)              # acotar rango ±20 %
    g2,_=pso(f,lo2,hi2,pop,iters,rng)
    ic_pso=0 if f.hist[-1]==f.hist[ev1-1] else ic_iter(f.hist,ev1,len(f.hist),pop)
    f.info=dict(EV_GWO=ev1,EV_PSO=f.ev-ev1,F1val_GWO=f1_gwo,F1val_PSO=1-f.best,IC_GWO=ic_gwo,IC_PSO=ic_pso)
    return g2,ev1

# %%
def metrics(m,Xte,yte):
    pr=m.predict(Xte); ps=m.predict_proba(Xte)[:,1]; tn,fp,fn,tp=confusion_matrix(yte,pr).ravel()
    return dict(F1=f1_score(yte,pr),Precision=precision_score(yte,pr),Recall=recall_score(yte,pr),
                FPR=fp/(fp+tn),AUC=roc_auc_score(yte,ps),MCC=matthews_corrcoef(yte,pr)), pr

rows=[]; preds={}
for r in range(N_REP):
    seed=100+r
    Xtv,Xte,ytv,yte=train_test_split(X,y,test_size=0.15,stratify=y,random_state=seed)
    Xtr,Xva,ytr,yva=train_test_split(Xtv,ytv,test_size=0.1765,stratify=ytv,random_state=seed)
    rng=np.random.default_rng(seed)
    def run(name,fn):
        t0=time.time(); f=Fit(Xtr,ytr,Xva,yva,seed)
        if name=='A0_default': m=default_model(seed).fit(Xtr,ytr); ev=0; p=None
        else:
            v=fn(f); m=make(decode(v),seed).fit(Xtr,ytr); ev=f.ev; p=decode(v)
        mt,pr=metrics(m,Xte,yte); T=time.time()-t0
        rows.append(dict(rep=r,metodo=name,EV=ev,**getattr(f,'info',{}),T_seg=round(T,1),F1_val_best=1-f.best if ev else np.nan,params=str(p),**mt))
        preds[(r,name)]=(pr,yte); print(r,name,round(mt['F1'],4),'EV',ev,round(T),'s')
    run('A0_default',None)
    run('A1_random',lambda f:min((rng.random(5) for _ in range(N_RAND)),key=lambda v:f(v)))
    def grid(f):
        g=[0.0,1/3,2/3,1.0] if not QUICK else [0.0,1.0]
        best=min(product(g,g,[0.2,0.8],[0.2,0.9],[0.2,0.9]) if not QUICK else product(g,g,[0.5],[0.5],[0.5]),key=lambda v:f(np.array(v)))
        return np.array(best)
    run('A2_grid',grid)
    run('A3_GWO',lambda f:gwo(f,np.zeros(5),np.ones(5),POP,IT_SOLO,rng)[0])
    run('A4_cascade',lambda f:cascade(f,POP,IT_CASC,rng)[0])
    pd.DataFrame(rows).to_csv('resultados/resultados_crudos.csv',index=False)
R=pd.DataFrame(rows)

# %%
summ=R.groupby('metodo')[['F1','Precision','Recall','FPR','AUC','MCC','EV','T_seg']].agg(['mean','std']).round(4)
summ.to_csv('resultados/resumen.csv'); display(summ)
etapas=R[R.metodo=='A4_cascade'][['EV_GWO','EV_PSO','F1val_GWO','F1val_PSO','IC_GWO','IC_PSO']].agg(['mean','std']).round(4)
etapas.to_csv('resultados/etapas_cascada.csv'); display(etapas)

# Pruebas de Wilcoxon pareadas (una cola) con corrección de Holm dentro de cada hipótesis
def wp(m1,m2,met):
    x=R[R.metodo==m1].sort_values('rep')[met].values; y=R[R.metodo==m2].sort_values('rep')[met].values
    try: return wilcoxon(x,y,alternative='less' if met=='FPR' else 'greater').pvalue
    except Exception: return 1.0

HIP={'H.G  A4 vs A0 (F1)':[('A4_cascade','A0_default','F1')],
     'H.E.1 GWO solo vs A0':[('A3_GWO','A0_default','Recall'),('A3_GWO','A0_default','F1')],
     'H.E.2 A4 vs GWO solo':[('A4_cascade','A3_GWO','Precision'),('A4_cascade','A3_GWO','FPR')],
     'H.E.3 A4 vs Random/Grid':[('A4_cascade','A1_random','AUC'),('A4_cascade','A1_random','MCC'),
                                ('A4_cascade','A2_grid','AUC'),('A4_cascade','A2_grid','MCC')]}
res=[]
for h,tests in HIP.items():
    ps=[wp(*t) for t in tests]; order=np.argsort(ps); adj=np.zeros(len(ps)); mx=0
    for k,i in enumerate(order): mx=max(mx,(len(ps)-k)*ps[i]); adj[i]=min(mx,1)
    for (m1,m2,met),p,pa in zip(tests,ps,adj):
        d=(R[R.metodo==m1][met].mean()-R[R.metodo==m2][met].mean())
        res.append([h,m1,m2,met,round(d,4),round(p,4),round(pa,4),pa<0.05])
W=pd.DataFrame(res,columns=['hipotesis','A','B','metrica','dif_media(A-B)','p','p_Holm','signif(0.05)'])
W.to_csv('resultados/wilcoxon_holm.csv',index=False); display(W)

# McNemar A4 vs A0 (última repetición)
r=N_REP-1; p4,yt=preds[(r,'A4_cascade')]; p0,_=preds[(r,'A0_default')]
b=int(((p4==yt)&(p0!=yt)).sum()); c=int(((p4!=yt)&(p0==yt)).sum())
stat=(abs(b-c)-1)**2/max(b+c,1); print('McNemar A4 vs A0: b=%d c=%d chi2=%.3f p=%.4f'%(b,c,stat,1-chi2.cdf(stat,1)))

# %%
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,2,figsize=(11,4))
R.boxplot(column='F1',by='metodo',ax=ax[0]); ax[0].set_title('F1 en prueba (por repetición)'); ax[0].set_xlabel('')
R.groupby('metodo')['T_seg'].mean().plot.bar(ax=ax[1],title='Tiempo medio (s)',rot=30)
plt.suptitle(''); plt.tight_layout(); plt.savefig('resultados/comparacion.png',dpi=200); plt.show()

# %%
# Tabla final de INDICADORES de las dos matrices (promedio de las 10 repeticiones) con su índice
def nivel(v,cortes,etiquetas):          # cortes ascendentes; etiquetas: una más que cortes
    for c,e in zip(cortes,etiquetas):
        if v<c: return e
    return etiquetas[-1]

A4=R[R.metodo=='A4_cascade']; A0=R[R.metodo=='A0_default']; A2=R[R.metodo=='A2_grid']
dF1=((A4.F1val_PSO-A4.F1val_GWO)*100).mean()
filas=[]

# Variable independiente (cascada A4)
v=A4.F1val_GWO.mean();  filas.append(['VI','F1-GWO (0-1)',round(v,4),nivel(v,[0.80,0.85],['Bajo','Medio','Alto'])])
v=A4.IC_GWO.mean();     filas.append(['VI','IC-GWO (iter.)',round(v,2),'Rápida' if v<=2 else 'Moderada' if v<=4 else 'Lenta'])
filas.append(['VI','ΔF1 PSO sobre GWO (pp)',round(dF1,3),'Sin aporte' if dF1<=0 else 'Marginal' if dF1<=0.5 else 'Relevante'])
v=A4.IC_PSO.mean();     filas.append(['VI','IC-PSO (iter.)',round(v,2),'Rápida' if v<=2 else 'Moderada' if v<=4 else 'Lenta'])
v=A4.EV.mean();         filas.append(['VI','EV (evaluaciones)',round(v,1),'Igual a Grid' if abs(v-A2.EV.mean())/A2.EV.mean()<=0.05 else 'Desigual'])
v=A4.T_seg.mean();      filas.append(['VI','T (s)',round(v,1),'Bajo' if v<120 else 'Medio' if v<=300 else 'Alto'])

# Variable dependiente: preprueba (A0) y posprueba (A4)
for nom,D in [('A0 preprueba',A0),('A4 posprueba',A4)]:
    for met,etq in [('Recall','R (%)'),('F1','F1 (%)'),('Precision','P (%)')]:
        v=D[met].mean()*100; filas.append(['VD '+nom,etq,round(v,2),nivel(v,[80,90],['Baja','Media','Alta'])])
    v=D.FPR.mean()*100; filas.append(['VD '+nom,'FPR (%)',round(v,3),'Baja' if v<2 else 'Moderada' if v<=5 else 'Alta'])
    v=D.AUC.mean();     filas.append(['VD '+nom,'AUC-ROC',round(v,4),nivel(v,[0.95,0.98],['Baja','Media','Alta'])])
    v=D.MCC.mean();     filas.append(['VD '+nom,'MCC',round(v,4),nivel(v,[0.70,0.85],['Bajo','Medio','Alto'])])

IND=pd.DataFrame(filas,columns=['variable','indicador','valor','indice'])
IND.to_csv('resultados/tabla_indicadores.csv',index=False); display(IND)
```
